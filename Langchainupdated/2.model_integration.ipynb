## Streaming and Batch

### Streaming
- streaming means generating the response chunk by chunk instead of waiting for the full response 
- It provides Immediate feedback to users
- Its good for chatbots and AI Assistants

```python
## Instaed of model.invoke() use model.stream()

response=model.stream('Tell me about India')
response
```

- So basically 'model.stream()' returns generator not complete response
- so what is meant by generator? The Answer is ,the generator is a object that produces values one at a time when you request them  

```python
## Generator Example
def my_generator():
  yield 'sai'
  yield 'durga'
  yield 'prasad'

result=my_generator()
result
```

```python
## So how to request them? ,Using 'for loop'

for item in result:
  print(item)
```

```python
## In the same way you request them to our generator that generated by LLM

for chunk in model.stream('Tell me about India'):
  print(chunk.content,end="",flush=True)

## end="" prevents Python from adding a newline after every chunk
## flush=True ,It helps display the text Immediately
```

### Batch

- Batching means giving multiple input and getting corresponding outputs from the model
- Instead of calling invoke() every time for each input ,we can use batch()

```python
## Batch Example
batch_responses=model.batch(
  [
    "Tell me about India",
    "Explain about the transformer",
    "Why parrot feathers are green?"
  ],
  config={
    'max_concurrency':5  #max_concurrency limits how many runnable executions can run concurrently.
  }
)

for response in batch_responses:
  print(response.content)
  print('------------------')
```
