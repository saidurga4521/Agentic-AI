### Messages
- Messages are the fundamentals unit of context for ai models
- They defined the input and output of the models
- Messages are object ,It contains 'role','content' and 'metadata'
  - Role :Identify the message(User,Sytem)
  - Content: Type of the content (text,video,image)
  - Meta data: Like toke usage and message ids
- Langchain provides standard messages to work across all the models ,Ensuring consistant behaviour regardless of the model

### Message Prompts
Messages prompts means you can give the multiple message types to the model
- System Message
- User Message
- Ai Message : It represents the what ai model returns.It contains normal text or request to call a function
- Tool Message: It contains the result returned by tool after calling 

```python
from langchain.messages import HumanMessage,AIMessage,SystemMessage
messages=[
  SystemMessage('you are an ondc export'),
  HumanMessage('explain about the ondc')
]

model.invoke(messages)
```

```python
### Problem :AI Shopping Assistant
### The user asks: "What is the total price of 3 notebooks if each costs ₹50? and also what is the gst on this price also"
# create the message
from langchain.messages import SystemMessage,HumanMessage,AIMessage,ToolMessage

## system message
sys_message=SystemMessage(
  content='Act like a Good mathematician to calculate the prices of products'
)

## human message
human_msg=HumanMessage(
  content='What is the total price of 3 notebooks if each costs ₹50?'
)
```

```python
# create a tool
from langchain.tools import tool
@tool
def calculate_total(price:int,quantity:int)->int:
  '''Calculate the total price of products'''
  return price*quantity

# create one more tool
# @tool
# def calculate_gst(gst:int,price:int)->float:
#   '''Calculate the gst on price'''
#   gst_val=(price//100)*gst
#   return gst_val
```

```python
# create the model
from langchain.chat_models import init_chat_model
from langchain_openai import ChatOpenAI
# model=init_chat_model('gpt-6-luna')
ai_model=ChatOpenAI(model='gpt-6-luna')
```

```python
# bind the model
model_with_tools=ai_model.bind_tools([calculate_total])
ai_msg=model_with_tools.invoke([
  human_msg,sys_message
])
ai_msg
```

```python
## get the tool calls from response
ai_msg.tool_calls
```

```python
## call the tool
tool_result=calculate_total.invoke(ai_msg.tool_calls[0]['args'])
print(tool_result)
```

```python
## Define the tool message
tool_msg=ToolMessage(
  content=str(tool_result),
  tool_call_id=ai_msg.tool_calls[0]['id']
)
```

```python
## create the final message for our required output
messages=[
  sys_message,
  human_msg,
  ai_msg,
  tool_msg
]

## invoke the model with this final message
final_result=model_with_tools.invoke(messages)
print(final_result)
```
